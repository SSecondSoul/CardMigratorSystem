## Observed interaction defect and repair
Decision buttons and note resolve/reopen/delete actions did not update the UI because the handlers directly mutated nested plain objects returned by `selectedAsset`. The v2 repair creates new assets and notes and writes the entire `assets` array through `this.data.set`. The same approach was applied to note addition.
