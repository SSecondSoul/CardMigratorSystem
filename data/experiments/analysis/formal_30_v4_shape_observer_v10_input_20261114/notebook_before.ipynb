## 六项未通过与评估器／候选门复核

5 项在第 1 次模型候选被拒后回退停止，不是每项都耗尽 3 次：StockReservationBadge 候选新增 available／warningAt 默认值遗漏；HabitWeekTracker 新增 habits／dayLabels 默认值遗漏并失去视觉通过；TripItineraryPlanner 新增 tripName／days／initialItems 默认值遗漏并失去视觉通过；TaskManager 虽从 10 条静态诊断降到 3 条，仍新增 TaskForm.isTagSelected 对 formData 的隐藏依赖，不能仅按总错误减少绕过保护。

RuleBuilder 有不同性质的门禁问题：保留件缺失 title、availableFields、initialGroups、readonly，候选仅缺 initialGroups。当前门禁以 code/location/message 为完整身份，整条分组文字改变后将缩小的旧缺陷记成新增。这个回退理由没有证明新字段缺失；但候选仍有默认值错误，不能因此直接判通过或全局忽略同码诊断。下一步应做基于源契约的逐字段结构化身份和新字段遗漏必须拒绝的反例。

MonitoringDashboard 使用 3 次模型修复后无进展，保留两条诊断需复核：simulateDataUpdate 实际用 data.get('kpiData').map(kpi=>({...kpi})) 克隆每个记录，本地四处 .value 标量修改之后再 data.set 写回。独立 AST 证明这些具体局部写入的克隆／提交形态；现有正则初始别名收集只匹配 data.get 前缀，产生误报。不代表所有 map／浅复制都安全，更不证明整个组件等价。

该组件持久化写入并未丢失，而在 try/catch 内。现有直接尾部写入证明拒绝 try/catch，记录 unsupported_callback_effects，避免盲补第二条写入；但诊断 missing trailing write 混淆了“缺失”和“异常包装未证明”。源未包装异常，异常传播／处理策略仍需另行审查，不能一概判完全等价。初始化契约 inited=0、attached=1 仍满足，方法接口无缺失，不是所有旧修复都失效。

新增 6/6 只读审计测试通过，3.892 秒，含真实全批账目、分组收缩、对象复制／身份 map／深层修改／提交顺序反例；不是重新完整跑 413 项。运行前 407 项结果仍独立保留。本次没有修改生产规则、Prompt、任何生成产物或测试页面，也没有新模型或浏览器操作。

## 评估校准 v1：实施边界（2026-11-04）

本节接续上节审计，记录已实施的通用评估校准，不是新的生成批次，也不回写原始 24/30。修改前重新验证结果包 457 项、运行协议 351 项 SHA-256；旧协议及其校验和不变。

1. 默认值诊断：ValidateStage 从 SSM props 默认值契约产生 details.contract=prop_defaults_v1 与 affected_fields；候选门按字段比较，不再依赖整条分组文字。缺失字段集合收缩可继续修复，但新字段缺失仍拒绝。旧记录缺少结构化信息时保守比较，不通过解析中文消息自动豁免。离线历史对照会重新验证两侧代码。
2. 克隆记录：新增通用 AST 证明模块 detached_record_semantics，仅认可完整可证明的局部使用链：map 创建新数组和逐记录对象展开副本、直接浅层标量修改、随后唯一顶层 data.set 写回。重赋值、引用逃逸、深层修改、未知调用、遮蔽、提前或缺失提交等均不获得豁免。它不是任意 map/浅复制安全的证明，也没有重写生成的 San。
3. 持久化：区分 source_watch_persistence_not_landed 与 source_watch_persistence_unproven。已有 try/catch、条件或不明确写入时保留错误，不补第二次写入、不自动拆除异常处理；原本缺失且满足完整源契约的直接尾写入仍可确定性恢复。新增诊断同步接入修复入口、语义回退、潜在诊断审查和聚焦 repair Prompt。

生产修改位于 stages/validate.py、stages/repair.py 和 utils 下的 repair_scheduler.py、repair_prompt.py、semantic_coverage.py、watch_persistence_semantics.cjs、detached_record_semantics.py/.cjs。generation 与本地服务器 Prompt 未改。新增 17 项自动测试（15 项合成/边界测试、2 项真实冻结产物测试），完整运行证据见后续结果节。

控制存储夹具验证：正常存储时直接写入与 try/catch 写入结果一致；存储抛异常时，源写法向外传播，包装写法吞掉异常，两者不同。因此没有将包装直接视为等价，也没有擅自扩大异常容忍策略。夹具不访问真实浏览器存储。

执行脚本：migration_pipeline/replay_formal_30_v4_evaluator_calibration_v1.py。逐项重新静态验证/编译冻结的 30 个保留产物，并对 5 个回退候选重新比较保护检查；视觉标志仅引用冻结报告，未重新执行视觉/交互评估。没有模型请求、没有覆盖标准 San、没有修改测试页面或旧分数。

## 源自定义事件单 payload 双端闭环 v1：实施（2026-11-10，部分修复）

### 已发现但未修复

- `NATIVE_DRAG_TRANSFER_SIDE_EFFECTS_MISSING`（production_defect）：冻结源 TaskItem.onDragStart(e) 先设置 text/plain 与 effectAllowed='move' 再发出索引；目标实际声明处理器遗漏前两项。相同模拟事件调用源函数与 San 声明可复现。当前 Validate 仍 0 错误，说明现有 DOM／数组事务规则并未覆盖该事件体副作用；浏览器依赖的原生拖拽影响未运行，不宣称所有浏览器必现。
- `PARENT_WATCH_REGISTRATION_AFTER_INITIALIZATION`（remaining_risk）：空存储、已保存数据、关闭自动保存、损坏 JSON 四场景，真实父应用 attached 钩子均先加载，再注册 tasks/settings；初始化监听回调与保存次数均为 0，显式统计正常。源是 mounted 前的声明式 deep watch，但本轮未执行 Vue runtime 异步队列，不把预计源回调次数写成实测。简单提前注册可能令 San 同步回调在恢复 settings 前保存或重启计时器，不能未经副作用证明就搬动。
- `SAN_DEEP_WATCH_NOT_VUE_DEEP_WATCH`（remaining_risk）：安装版 San watch 只有两个形参，第三个 deep/immediate 选项不生效。嵌套 data.set、data.splice 会触发；原对象直接突变及同引用 set 不触发。保留源深监听兼容性风险，不视为已形成通用闭环。

## 父监听：真实 Vue 异步队列对照 v1（2026-11-11）

本轮按上一节计划先建立运行 oracle，**未实施父监听修复**。记录技能要求将实测、未修复状态与人工验收分开，故不把诊断测试通过写成组件已修好，不增加批次通过数。

### 隔离运行依据

本地仅有 Vue 编译器，无缓存 Vue2 runtime。沙箱 npm 元信息请求因 EACCES 失败；随后经授权直接读取官方 npm 元信息／下载 Vue **2.7.16** 发布包，核对 dist SHA-512 与 SHA-1，再只提取 runtime、MIT LICENSE、package.json。包与精确运行字节、来源记录均冻结；不安装依赖／执行安装脚本，不改项目 package／lock／node_modules。测试页 `vue@2` 未锁补丁，不能认定本 oracle 就是浏览器当时版本。实际使用安装版 San 3.15.6，不代表人工页 3.15.3。

官方 [异步更新队列](https://v2.vuejs.org/v2/guide/reactivity.html#Async-Update-Queue)、固定版 [initWatch](https://raw.githubusercontent.com/vuejs/vue/v2.7.16/src/core/instance/state.ts) 与 [scheduler](https://raw.githubusercontent.com/vuejs/vue/v2.7.16/src/core/observer/scheduler.ts) 用于核对实现背景；下列结论来自运行固定版真实框架，而非自写 Vue 队列。

5 存储场景 × 3 对照（冻结源 Vue／当前 San／仅在内存提前注册的 San）= **15 次执行**，每次 18 个同步／nextTick 阶段，共 **270 份状态／事件快照**。实际 Vue 构造、声明式深 watch、默认 async、根 $mount→mounted、$destroy→beforeDestroy；render 为无子组件的空根 VNode，Node patch 无 DOM，不覆盖源模板。San 实际构造／inited／内建 watch／nextTick／原生 dispose，attached 应用钩子直接调用，未 DOM attach。独立 vm 沙箱，固定时钟，模拟存储和应用 interval/timeout，但实际框架 Promise 队列未用手写调度器替代。未触发周期云同步或 3000ms 通知回调。

### 初始化保存次数

| 场景 | 源 Vue | 当前 San | 仅提前注册 San |
| --- | ---: | ---: | ---: |
| 空存储 | 1 | 0 | 1 |
| 已保存、手动同步 | 2 | 0 | 2 |
| 已保存、autoSave=false／30 秒 | 0 | 0 | 1（错误默认设置） |
| 已保存、autoSave=true／30 秒 | 2 | 0 | 2（首份载荷仍为默认 60 秒） |
| 损坏 JSON | 1 | 0 | 1 |

**简单提前注册不安全**：源 Vue 在 mounted 同步恢复设置后才执行队列；提前注册的 San tasks 回调同步发生在 settings 恢复前。autoSave=false 的存储会先被写成 autoSave=true、syncInterval=60，后续 settings 恢复为 false 又不再保存，模拟存储最终留在错误默认值。此为未应用的内存反事实，不计为已实施生产修复回归。

**调度也有差异**：同轮两次标题修改，源 tasks 监听运行 1 次，San 运行 2 次。先改 settings 两次再改 tasks，源仍按注册次序 tasks→settings 各一次；San 为 settings→settings→tasks，写中间态并重复重启 interval。保存 30 秒间隔的源初始化确实启动一次、由 settings watcher 清理重启一次，需保留其实际效果，不能擅自优化。

**销毁边界**：改设置后立刻销毁，源待排 watcher 不再执行，beforeDestroy 保存 1 次；San 已同步回调，保存和建 timer，再 disposed 保存，共 2 次。真实框架销毁后再改数据均无监听／保存，活动 interval 为 0；源本来未取消的通知 timeout 仍在模拟表中，未触发，不能宣称所有 timer 已清理。

各 tick 后任务、设置、统计值仍相同，**最终画面／最终值相同不代表副作用历史等价**。此前“父监听时机 remaining_risk”现升级为 **1 类已运行复现的迁移缺陷**（初始化遗漏／去重／顺序／销毁前效果同属监听调度缺口，不按症状重复计数）。这不是本轮原生拖拽或工厂修复引入的回归，也不能继续仅归为模型随机。

### 冻结与验证

## 父默认异步监听有界闭环 v1（2026-11-11）

本节承接上一节真实 Vue 队列 oracle，实施通用但有严格前提的修复，不再只移动 watch 或追加 Prompt。证据：[REPORT.md](../../data/experiments/analysis/formal_30_v4_parent_watch_closure_v1_20261111/REPORT.md)、[replay.json](../../data/experiments/analysis/formal_30_v4_parent_watch_closure_v1_20261111/replay.json)、[unit_tests.txt](../../data/experiments/analysis/formal_30_v4_parent_watch_closure_v1_20261111/unit_tests.txt)。31 项 SHA-256 payload，382 项运行依赖清单。独立完整测试 **672/672、0 失败、0 跳过，1238.068 秒**；不与前期局部测试相加。

### 问题、根因与通用修复

上节运行时复现的 1 条父监听迁移缺陷（production_defect）在此修复：源 Vue 在 mounted 前注册、默认异步合并并按声明顺序执行 deep watch；目标 San 在 attached 初始化后才注册、同步执行，造成初始化漏保存、重复保存／重启 interval、顺序与销毁前效果不一致。这不是原生拖拽修复引入的回归，也不能归为模型随机。

新增 `migration_pipeline/utils/source_parent_watch_semantics.py/.cjs`。只有纯初始 data、默认异步根字段 deep watch、零参同步回调，以及完整 mounted/beforeDestroy 和传递效果方法均能经 AST 对应时，才自动改写。每实例闭包队列在 inited 注册、按源声明顺序去重，Promise 微任务执行时读取最新状态，disposed 取消尚未运行任务；保留源正常回调的全部业务效果。拒绝未知阶段／外部依赖／mixins／特殊 sync、immediate 选项／递归／监听回调写回被监听根字段等，不能证明时保留诊断，不覆盖未知代码。组件名、字段名不特判，含改名合成用例。

真实对照还发现同值入口：源 `settings[key]=value` 对已存在键的同值赋值不通知，目标克隆整个对象会错误触发监听。仅当完整源赋值、目标四段克隆写回和后续效果精确对应，才加入 `Object.is` 同值保护；后续通知仍保留。这里只证明已声明现有键，不宣称任意未知／新增／原型键的 Vue2 非响应式语义已等价。

修改位置：`stages/repair.py` 的 residual_v2 确定性前处理和模型返回后处理、`utils/repair_prompt.py` 的共享修复约束、`stages/validate.py` 的两类 source_parent_watch 诊断、`utils/repair_scheduler.py` 的契约丢失／语义回归拒绝。generation 和服务器 Prompt、legacy 前处理未改；不手改 canonical San，不切测试页面。旧测试的原有断言保留，并补新诊断及组合规则预期。

### 实测方法与发现

对上轮冻结的 TaskManager 派生产物和冻结源 Vue，执行 **9 场景×2 框架×10 阶段＝18 次运行、180 份快照、90 组对照**。源 Vue2.7.16，San3.15.6；源业务方法、目标代码均不改写。应用存储和 timer 为隔离模拟，只有一次 localStorage.setItem 故障注入；原生 Promise 队列真实运行，两个宿主 setImmediate 仅等候队列与拒绝通知。无浏览器／源模板／子组件／San DOM attach，San 应用 attached 直接调用、真实 headless dispose。未触发全部应用 timer，也不宣称全部 timeout 已清理。

三个控制场景（声明键同值、声明键变化、恢复对象中已有的额外键）全阶段状态与业务效果匹配；这也说明不能仅以“是否出现在初始 data 字面量”判断某键有无响应式 setter。余下 6 个边界场景用于记录差异，不是 6 个独立组件缺陷，也不能算作 30 样本的新通过率。

- **PARENT_WATCH_DYNAMIC_KEY_SHAPE_NOT_CLOSED**（remaining_risk，TaskManager，运行时已复现、未修复）：事后新增普通键／继承键，Vue2 直接赋值不触发 deep watch，而目标克隆整个 settings 会额外保存、重启 interval。新增值为 undefined 时，源会创建自有属性，目标 Object.is 同值保护却跳过，属性本身缺失。__proto__ 在源中改变局部 settings 原型，在 spread 目标中形成自有属性。对照同时记录自有可枚举键、undefined 键、原型身份与继承标记，不能只靠 JSON.stringify 忽略这些差别。原型输入限于独立 vm 内的 settings；宿主及沙箱 Object.prototype 均未污染。
- **PARENT_WATCH_EXCEPTION_POLICY_NOT_CLOSED**（remaining_risk，TaskManager，运行时已复现、未修复）：显式配置 Vue errorHandler 记录一次存储写入失败后，tasks 回调中止，但 settings 回调继续保存并启动 30 秒 interval；当前 San 派生队列整个 flush 中止，后一个监听未运行，错误为 unhandledRejection。San 下轮又能正常运行，说明 finally 复位有效，但不能据此称错误策略等价。另一个无自定义 handler 的 Node 对照中，Vue 抛错后后续监听队列不再运行，San 则恢复；不能通用地“catch 后继续”假装对齐两种策略。显式 capture 是装置配置，**不是浏览器默认策略证据**；不手写或模拟浏览器环境来凑结论。

宿主 unhandledRejection 监听只观察错误，不给被测 Promise 附加 catch，不更换调度器，结束时移除。错误通道单列；业务方法、保存尝试／成功／失败、timer 启停全部保留。后续 timer ID 不同可源自先前缺失启动，不另算一条缺陷。控制与边界都检查恢复和销毁，而非只看最终画面。

本步完成的是**防止未证明分支被自动修成通过的保护闭环**，不是异常策略或任意动态键语义已修好。证据目录按工作开始日期命名，实际冻结记录 `created_at=2026-11-12T00:06:01+08:00`。证据：[REPORT.md](../../data/experiments/analysis/formal_30_v4_parent_watch_scope_guard_v1_20261111/REPORT.md)、[replay.json](../../data/experiments/analysis/formal_30_v4_parent_watch_scope_guard_v1_20261111/replay.json)、[unit_tests.txt](../../data/experiments/analysis/formal_30_v4_parent_watch_scope_guard_v1_20261111/unit_tests.txt)。25 项 SHA-256 payload、387 项运行清单；最终独立完整测试 **703/703、0 失败／跳过，1117.584 秒**，不与之前的完整或局部测试相加。

### 实施的通用保护

`source_parent_watch_semantics.py/.cjs` 独立检查源 watcher 传递效果链和源 methods 的对象写入，不等待目标方法先被识别成功。效果链访问 localStorage，输出 `source_parent_watch_exception_policy_unproven`；对被监听初始对象存在未限定计算键、静态新增／继承／原型键或 delete／解构写入等已识别边界，输出 `source_parent_watch_dynamic_key_unproven`。已有 try/catch 也不被直接当作完整运行时错误适配器。这是保守收紧，不是证明所有 JavaScript 异常、间接别名、动态反射或任意对象形状均已处理。

有这些边界则不自动生成新队列或同值 clone 保护；旧正常路径队列可保留 `landed=true` 的局部事实，但同时标为 `proof_scope=normal_flow_only`、`automatic_repair_eligible=false`，两项错误仍保留。纯非存储、已知静态形状组仍能规范化且保持不动点；字段／方法改名、传递调用、写入前有其他语句、缺失目标方法等均有测试。

验证器输出明确建议；共享 repair Prompt 禁止臆造 catch-and-continue、吞异常、删未知键或强迫 Vue2 新增属性响应式。repair.py 既有前／后处理因调用此规则而遵守拒绝条件，未改生成／服务器 Prompt，也未扩展 legacy 入口。bounded 调度器保留可证明的其他规范化，随后在模型调用前停止为 `source_parent_watch_scope_requires_review`；现行候选保护独立重算源边界，伪造通过／删除诊断会被 `waived_source_parent_watch_boundaries` 拒绝，但不拒绝仍如实报告边界的无关进步。

### 结果与失败经过

30 个累计冻结产物均完成现行静态／编译回放，**全部原字节未改写**；原 24 个通过产物仍静态通过。原批次仍 **24/30、6 失败、0 执行异常、1,821,544 tokens**。本轮不重做初始 DOM，只引用旧视觉结果，不继承浏览器／人工验收；Trip 的外部渲染依赖及其他既有诊断保留，不改成新批次通过率。

TaskManager 的冻结正常路径产物，从历史静态 0 错误变为现行 **2 项边界错误**：这是同一产物被更严格判定，不是新代码回归。正常路径已修的输入、尚未补父 watch 的输入都经真实离线 bounded 调度器验证：**0 规范化步、0 LLM、代码不变、final_passed=false、明确待复核**。这里“待复核”指源／运行语义契约的工程复核，不能用页面人工点验通过直接清除诊断。

冻结边界装置再次执行 **18 次、180 快照、90 对**，观察与上节完全相同；差异仍在，证明未通过改变业务、吞错误或过滤副作用凑通过。它仍是固定 Vue2.7.16／San3.15.6、模拟存储／应用 timer、真实 Promise 队列、无浏览器／完整 DOM attach 的证据。

开发预检中的失败单列保留：新增目标方法缺失测试最初用不匹配空格的字符串，未实际改变输入而失败；修正后验证。第一次完整运行 **702 项、1237.042 秒、2 失败**：旧子组件调度用例仍期待全通过，旧 TaskManager 回退用例期待进入模型阶段，而新策略已更早停止。最终更新为验证提前停止，并新增无此边界的合成组件、使用真实验证／视觉阶段继续验证原文件恢复、拒绝文件及历史证据保留，**不删除回退保护、不屏蔽生产诊断、不放宽规则**。首次完整 stdout 未落盘，冻结数据明确它只是工具输出摘要；最终独立完整 703 项日志另存，不能把两次运行拼起来。

本轮模型请求 0；不改 canonical San 或测试页，不提交／推送。按记录技能将“保护闭环完成”与“语义修复未完成”分开。仅追加，前 **48 个单元及 metadata 原样保留**；历史实时指纹检查的预期版本漂移仍保留。下面只核对冻结记录并重放原边界装置，未运行 Jupyter UI，不导入当前生产规则冒充历史实现。

## 2026-11-12：公开写入接口与历史设置数据审计（候选仍未部署）

## 2026-11-13：显式应用错误报告器——隔离候选v5，不代表默认异常闭环

按 **vue-san-validation-report** 保留错误通道、负向结果及证据边界。本轮只完成一个明确约定：应用显式提供、同步返回、仅记录错误而不读写组件状态的报告器。没有替现有应用选定新的异常策略，没有修改生产Prompt/规则、旧探针、canonical San或测试页面。

根因：Vue对用户watcher逐个执行错误处理；合并Promise队列若不在单个dispatch处处理异常，一个存储写入失败会退出flush，使后续watcher漏执行。v5仅增加一个dispatch catch和一个实例配置方法；收到原Error、当前组件、精确watcher信息，使用null作为报告器this。报告器返回后继续下一个watcher，但不恢复已失败回调内部尾部；不重试存储、不加默认日志或Promise.catch。报告器缺失时抛原异常，报告器抛错也向外传播，均不冒充Vue全局回退已实现。AST还原检查证明其余脚本（v4观察桥、赋值右侧、业务方法、队列/销毁等）不变。

原生Vue2.7.16/San3.15.6，9场景×3变体×10阶段=27运行、270快照、180比较。6个明确报告约定场景中，源/v5共60/60阶段的状态、完整业务/应用错误报告事件、警告匹配；检查同轮双错误各报一次、tasks失败后settings仍执行、settings失败不执行自身后续同步、后轮失败保留旧interval、返回false不误终止，以及原异常身份、组件归属、watcher信息和先后顺序。原v4无报告器对照仍保留。此装置为headless原生框架/Promise，存储和业务timer是故障注入模型，不是真实浏览器或真实存储验证。

**三个负向组没有修复，也没有删除：** Node默认策略、目标报告器缺失、报告器自身抛错，源/v5分别仍有7、5、7个差异阶段（合计19/30）。Node默认源调度器抛错后的后续状态不同，不能只比较第一次报错；报告器自身抛错的全局回退、祖先errorCaptured、异步/可重入报告器、依赖收集及浏览器默认路由仍未证明。rawErrors完整保留Vue.config.errorHandler与San实例报告器的不同通道，不把框架标签硬改为相同。

此前冻结9场景装置不配置报告器时，v5的全部观察与v4完全一致，两个历史存储异常失败仍在；非抛错原生子组件装置的全部观察也与v4一致。报告器配置是隔离候选的应用适配入口，不是现有应用已部署功能，更不是缩窄源公开API或删除生产诊断。候选San编译通过，生产验证仍失败；动态键/异常策略边界及伪造通过保护均保留。

本地页面静态核对：Vue runner脚本未显式声明Vue.config.errorHandler，引用CDN为@2浮动主版本；San页引用3.15.3，而探针使用固定2.7.16/3.15.6。验收脚本的window error/unhandledrejection监听只记录事件，不构成watcher恢复策略。未联网取CDN或启动页面，不推断浏览器实际全局配置，不修改页面。这也是不能把本次显式报告约定自动当作应用默认策略的原因。

证据：[REPORT.md](../../data/experiments/analysis/formal_30_v4_watch_error_router_v5_20261113/REPORT.md)、`audit.json`、`scoped_tests.txt`、隔离派生San及`SHA256SUMS.txt`。57/57定向测试（20新增约定+23既有v4+14生产守卫），0失败/0跳过，非完整703项重跑；旧409项清单（含387项生产/既有运行文件）、30个保留输入/Vue、两个页面和旧62单元/元数据保留。新增4文件后413项，9个payload封存。

## 2026-11-13：原生浏览器分支、祖先捕获与全局回退——负向审计

按 **vue-san-validation-report** 记录新增负向证据，本轮不是修复成功，也不新增生产缺陷率。使用原封存v5候选原字节；不修改适配器、生产Prompt/规则、canonical San、两个测试页面或验收脚本，不调用模型。

固定Vue2.7.16/San3.15.6在旧封存的受限miniDOM装置中实际创建并挂载Far→Near→Subject，验证真实`$parent`/`parentComponent`身份；Vue真实执行`inBrowser`分支。Subject仍是封存TaskManager的业务方法/生命周期，仅替换根render/template为最小节点；祖先hooks/global handler是测试应用环境，不声称现有应用已经配置。无私有错误函数调用，原生Promise实际执行，Node宿主只观察拒绝。**这不是Chrome等真实浏览器、CSS、真实存储/业务timer或canonical业务UI验收。**

11场景×2变体×10阶段=22运行、220快照、110对照。正常与显式返回global handler两组共20/20阶段在状态、事件、原始console及拒绝上完全匹配；其余9组仍存在差异，完整原始Vue warning组件栈、Error严格身份、hook顺序和Promise拒绝均保留。9组是同一未闭合错误上下文契约的场景覆盖，不是9条新生产缺陷。

源约定已复现：浏览器默认无handler时记录原异常后仍执行下一watcher；errorCaptured按由近到远、同祖先声明顺序调用，**返回false才停止传播，返回true不停止**。钩子this是捕获祖先，错误subject仍是原组件。捕获钩子自身抛错时，次生错误以该祖先及`errorCaptured hook`归属进入全局路径，原错误之后仍可继续被远祖捕获。

全局handler抛不同错误时，Vue浏览器分支先记录次生错误、再记录原错误并继续队列；重新抛原错误只记录一次。v5没有该fallback，也不能在无报告器时等价地走浏览器默认日志路径。两组`capture_returns_true_global`与`near_throw_global_returns`所有业务状态阶段相同，但hook路径/事件仍不同，说明仅看最终UI/状态会漏掉错误路由差异。

每个San主运行另做一个原生watch阳性控制，共11个（不混入上述220快照分母）：原Error经实际父链到达最近ancestor.error，故钩子本身可用。然而最近error返回true时，San仍停止，不向远祖继续传播；主自定义Promise队列异常又绕过了这些已安装的San祖先钩子。**不能把Vue errorCaptured简单改名为San error，也不能仅给队列统一catch就声称等价。**

证据：[REPORT.md](../../data/experiments/analysis/formal_30_v4_browser_error_chain_audit_v1_20261113/REPORT.md)、`audit.json`、`scoped_tests.txt`及`SHA256SUMS.txt`。54/54定向测试（20新审计+20既有v5+14守卫），0失败/0跳过；通过表示反例被复现/锁定，不表示候选异常策略已修复。未重跑完整703项，沿用原v5编译通过/生产验证失败的记录，不冒称本轮重新编译。生产动态键/异常边界继续保留。

## 2026-11-13：隔离版源错误上下文桥 v6（非生产放行）

在上一节冻结的浏览器/祖先链反例上新增独立helper、AST适配探针和原生运行装置；不改生产生成/修复Prompt、repair.py、测试页面或canonical San。根因是自定义合并Promise队列缺少源祖先捕获/全局回退上下文，且San最近error钩子不能等同于Vue errorCaptured。

**限定的应用约定**：必须显式提供native父链、捕获hook列表、全局handler getter和日志sink；profile限定同步、只报告、无响应式读写。近到远及声明顺序调用hook，仅严格false短路；捕获hook抛错归属该祖先并走全局，然后继续原错误；全局抛不同错误先记次生再原错误，重抛原错误去重。AST还原只移除新增工厂、配置方法、dispatch分支，其余v5脚本完全相同，不改源赋值右侧/业务/生命周期或全局San错误语义。
