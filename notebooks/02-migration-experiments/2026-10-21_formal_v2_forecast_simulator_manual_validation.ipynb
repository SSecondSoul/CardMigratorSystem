## Observed defect and formal_30_v2 repair
The San component had no initial simulation result, chart, summary values, or monthly detail rows. The final third-round v1 artifact had lost the `attached()` and `disposed()` hooks that earlier artifacts still contained, so Vue `mounted()` behavior no longer called `runSimulation()`. Its `initData()` also used Vue-style `this.initialAssumptions`, which is undefined in San initialization. This identifies a repair regression rather than an omission shared by every generated version.

The v2 copy restores `inited()` to clone `initialAssumptions` through `this.data`, `attached()` to run the initial simulation, and `disposed()` to clear the timer. The frozen v1 artifacts remain unchanged.
