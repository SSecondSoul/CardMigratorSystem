# formal_30_v2: InventoryBatchManager manual-runtime repair

## Automated baseline
formal_30_v1 stopped after three repair rounds without passing. The final parent artifact is retained as the source for this targeted manual repair.

## Observation
A single row selection did not change its row style, and the select-visible action appeared ineffective in the San browser page.

## Root cause
The template called `isSelected(batch.id)` and `batchRowClass(batch)` without passing `selectedIds`. San could update the data field, but these ordinary method calls did not expose `selectedIds` as template dependencies for recomputation.

## Fix
The v2 artifact passes `selectedIds` explicitly: `isSelected(batch.id, selectedIds)` and `batchRowClass(batch, selectedIds)`. The helper methods consume that argument.
