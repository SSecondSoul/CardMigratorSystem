## Observed defect and formal_30_v2 repair
Manual validation found that toggling a device time slot changed its text and dependent values, but its button color did not change. The class expression called `slotClass(device.id, hour)` and read `schedule` only inside that ordinary method, so San did not track `schedule` as a template dependency.

The v2 repair passes `schedule` explicitly as `slotClass(device.id, hour, schedule)`. The method derives the class from that argument, causing the class to be recomputed whenever the schedule changes. The frozen v1 artifact remains unchanged.
