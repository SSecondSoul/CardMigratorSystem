# TaskManager：单轮聚焦语义修复探针 v1

此探针以 `formal_30_v4_semantic_focus_v2` 冻结失败产物为输入，仅调用一次修复模型；不是 Vue 源码到 San 的重新生成，也不是独立正式测试。输入、Prompt 和实现预冻结于 `data/experiments/snapshots/formal_30_v4_task_manager_semantic_probe_v1_pre_run_20261028/`。结果冻结于 `data/experiments/snapshots/formal_30_v4_task_manager_semantic_probe_v1_result_20261028_155408/`，7 个 payload 的 SHA-256 校验通过。


## 结果

聚焦 Prompt 确实触发，长度 43,814 字符，Qwen-plus 单次调用 23,432 tokens。候选代码的静态校验仍失败，视觉评估通过（结构相似度 0.9889）；没有进行浏览器人工验收。输入的 `invalid_san_computed_instance_method_call`、`hidden_template_reactive_dependency`、`nested_state_mutation` 在报告中消失，但 `invalid_san_event_handler_syntax` 和 `lifecycle_behavior_not_landed` 仍存在，并新增 `missing_event_handlers`、`missing_event_bindings`、`event_arguments_not_landed`。不能以消失的错误码数量单独宣布修复有效。


## 失败机制与处理决定

探针产物把部分源事件处理器改名为 `handle...`，导致源处理器及参数绑定无法被校验器追踪；内联模板仍残留 Vue 风格 `.stop` / `.prevent` 事件属性。该后验候选不得标记为自动迁移通过，也不得改写 v2 的失败 checkpoint。下一轮应先加入源事件绑定守恒的回归门槛，并对事件语法／生命周期残留做更窄的修复，而不是再次进行高成本的完整生成复跑。审计记录：`data/experiments/analysis/formal_30_v4_generalization_gap_audit_20261028/task_manager_semantic_probe_v1_review.json`。
