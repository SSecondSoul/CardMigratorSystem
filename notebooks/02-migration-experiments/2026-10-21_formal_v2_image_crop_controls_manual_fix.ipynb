# formal_30_v2: ImageCropControls manual-runtime repair

## Observation
The San component had no initial crop box or initial numeric values in the browser, while the Vue component initialized the crop from `initialCrop`.

## Root cause
The San artifact initialized `crop` as an empty object but did not copy `initialCrop` during component initialization. The Vue source performs this work in `created()`.

## Fix
Added an `inited()` hook to copy `initialCrop` into `crop` with `Object.assign({}, initialCrop)`. The formal_30_v1 artifact and Vue source remain unchanged.

## Follow-up browser issue
After the initial-value fix, changing a numeric parameter did not move or resize the canvas crop box. The handler used `event` instead of San's `$event`, so `update()` did not receive the DOM event. The handler was changed to `update(field, $event)`.

## Lock-aspect-ratio follow-up
The checkbox row inherited the numeric-field grid style, placing its content incorrectly. In addition, the migrated Vue-style `watch` object was not registered by the San runtime, so enabling the lock did not recalculate height. A dedicated `lock-control` flex style was added, and the watcher was registered with `this.watch('locked', ...)` inside `inited()`.

## Manual validation
The user confirmed the initial crop box, live canvas updates, lock behavior, layout, and reset interaction. Manual result: passed on 2026-10-21.
